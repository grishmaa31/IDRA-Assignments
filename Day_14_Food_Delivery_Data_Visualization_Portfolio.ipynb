Day 14 - Food Delivery Data Visualization Portfolio

This notebook uses Matplotlib and Seaborn to explore food delivery business performance through different visualizations and interpret the patterns found in the data.

2. **Import libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

3. **Load the Dataset**

```python
df = pd.read_csv("Day14_Food_Delivery_Visualization_Dataset.csv")
df.head()
```

4. **Basic Dataset Information**

```python
print("Shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

5. **Line Plot — Orders over Time**

```python
df["Date"] = pd.to_datetime(df["Date"])

daily_orders = df.groupby("Date")["Orders"].sum()

plt.figure(figsize=(12, 5))
plt.plot(daily_orders.index, daily_orders.values, label="Orders")
plt.title("Orders Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Orders")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The line plot shows how the number of orders changes over time. It helps identify periods with higher or lower order activity and gives an overall view of ordering trends.


6. **Bar Chart — Revenue by City**

```python
city_revenue = df.groupby("City")["Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=city_revenue.index, y=city_revenue.values)
plt.title("Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The bar chart compares revenue across cities and makes it easy to identify which locations contribute more to the overall business revenue.


7. **Bar Chart — Revenue by Cuisine**


```python
cuisine_revenue = df.groupby("Cuisine")["Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=cuisine_revenue.index, y=cuisine_revenue.values)
plt.title("Revenue by Cuisine")
plt.xlabel("Cuisine")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The chart shows differences in revenue across cuisines and helps identify the cuisines performing better in terms of overall revenue.

8. **Scatter Plot — Marketing Spend vs Revenue**

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Marketing_Spend",
    y="Revenue"
)

plt.title("Marketing Spend vs Revenue")
plt.xlabel("Marketing Spend")
plt.ylabel("Revenue")
plt.show()
```

### Interpretation

The scatter plot helps examine whether higher marketing spending is associated with higher revenue. The overall direction and spread of the points give an idea of the relationship between the two variables.

9. **Scatter Plot — Delivery Time vs Customer Rating**

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Avg_Delivery_Minutes",
    y="Customer_Rating"
)

plt.title("Delivery Time vs Customer Rating")
plt.xlabel("Average Delivery Time (Minutes)")
plt.ylabel("Customer Rating")
plt.show()
```

### Interpretation

The plot helps examine whether delivery time is related to customer ratings. It can show whether longer delivery times tend to be associated with lower or higher ratings.

10. **Histogram — Revenue Distribution**



```python
plt.figure(figsize=(8, 5))
sns.histplot(df["Revenue"], kde=True)

plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

### Interpretation

The histogram shows how revenue values are distributed and whether most observations are concentrated around a particular range.



11. **Box Plot — Orders by City**

```python
plt.figure(figsize=(10, 5))
sns.boxplot(
    data=df,
    x="City",
    y="Orders"
)

plt.title("Order Distribution Across Cities")
plt.xlabel("City")
plt.ylabel("Orders")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The box plot compares the distribution of orders across cities and also helps identify differences in spread and possible unusual values.

12. **Violin Plot — Revenue by Cuisine**

```python
plt.figure(figsize=(10, 5))
sns.violinplot(
    data=df,
    x="Cuisine",
    y="Revenue"
)

plt.title("Revenue Distribution by Cuisine")
plt.xlabel("Cuisine")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The violin plot shows both the distribution and density of revenue for different cuisines, making it easier to compare how revenue values are spread.

13. **Count Plot — Order Channels**

```python
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x="Order_Channel"
)

plt.title("Orders by Order Channel")
plt.xlabel("Order Channel")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.show()
```

### Interpretation

The count plot shows how frequently different order channels appear in the dataset and helps identify the more commonly used channels.

14. **Count Plot — Weather**

```python
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x="Weather"
)

plt.title("Orders Across Weather Conditions")
plt.xlabel("Weather")
plt.ylabel("Number of Records")
plt.show()
```

### Interpretation

The chart compares the number of recorded orders under different weather conditions and helps understand how frequently each condition occurs in the dataset.

15. **Weather vs Revenue**

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="Weather",
    y="Revenue"
)

plt.title("Revenue Across Weather Conditions")
plt.xlabel("Weather")
plt.ylabel("Revenue")
plt.show()
```

### Interpretation

The box plot compares revenue across weather conditions and helps identify whether revenue levels and their spread differ depending on the weather.

16. **Correlation Heatmap**

```python
plt.figure(figsize=(12, 8))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

### Interpretation

The correlation heatmap provides an overview of relationships between numerical variables. Stronger positive or negative values indicate stronger relationships, while values closer to zero indicate weaker relationships.

17. **Revenue and Orders by Month**

```python
monthly = df.groupby("Month")[["Orders", "Revenue"]].sum()
monthly
```

```python
plt.figure(figsize=(10, 5))
plt.plot(monthly.index, monthly["Revenue"], marker="o", label="Revenue")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.legend()
plt.show()
```

### Interpretation

The monthly revenue plot shows how business revenue changes across the months and helps identify stronger and weaker periods.

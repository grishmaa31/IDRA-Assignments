1. **Title**

Day 11 - Cleaned Company Employee Dataset

This notebook identifies and fixes missing values, inconsistent entries, incorrect data types, and duplicate records using Pandas.

2. **Import Pandas**

```python
import pandas as pd
```

3. **Load the Dataset**

```python
df = pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv")
df.head()
```

4. **Check the Dataset Before Cleaning**

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

5. **Fix Inconsistent Text Entries**

```python
text_columns = ["Department", "Gender", "City", "Work_Mode"]
for column in text_columns:
    df[column] = df[column].str.strip().str.title()

print(df["Department"].unique())
print(df["Gender"].unique())
print(df["City"].unique())
print(df["Work_Mode"].unique())
```

6. **Handle Missing Values**

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Annual_Salary"] = df["Annual_Salary"].fillna(df["Annual_Salary"].median())
df["Experience_Years"] = df["Experience_Years"].fillna(df["Experience_Years"].median())
df["Performance_Score"] = df["Performance_Score"].fillna(df["Performance_Score"].median())
```

```python
df["Department"] = df["Department"].fillna(df["Department"].mode()[0])
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
df["City"] = df["City"].fillna(df["City"].mode()[0])
df["Work_Mode"] = df["Work_Mode"].fillna(df["Work_Mode"].mode()[0])
```

7. **Convert Date Column**

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
print(df.dtypes)
```

8. **Remove Duplicate Records**

```python
df = df.drop_duplicates()
print("Duplicate rows:", df.duplicated().sum())
```

9. **Check the Dataset After Cleaning**

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print(df.isnull().sum())
```

```python
df.info()
```

10. **Before vs After Comparison**

```python
print("Before Cleaning")
print("Rows:", 157)
print("Duplicate Rows:", 7)

print("\nAfter Cleaning")
print("Rows:", df.shape[0])
print("Duplicate Rows:", df.duplicated().sum())
print("Missing Values:", df.isnull().sum().sum())
```

12. **Export the Cleaned Dataset**

```python
df.to_csv("cleaned_company_employee_dataset.csv", index=False)
print("Cleaned dataset saved successfully!")
```

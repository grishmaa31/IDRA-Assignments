Day 9 - Processed E-commerce Dataset

This notebook combines Orders, Customers, and Products datasets using Pandas and creates a clean processed dataset.

2. **Import Pandas**

```python
import pandas as pd
```

3. **Load the datasets**

```python
orders = pd.read_csv("Day9_Orders.csv")
customers = pd.read_csv("Day9_Customers.csv")
products = pd.read_csv("Day9_Products.csv")

print("Orders:", orders.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
```

4. **View the datasets**

```python
orders.head()
```

```python
customers.head()
```

```python
products.head()
```

5. **Merge Orders and Customers**

```python
data = pd.merge(orders, customers, on="Customer_ID")
data.head()
```

6. **Merge with Products**

```python
data = pd.merge(data, products, on="Product_ID")
data.head()
```

7. **Demonstrate concat()**

```python
sample1 = orders.head(3)
sample2 = orders.tail(3)
combined = pd.concat([sample1, sample2])
combined
```

8. **DateTime Operations**

```python
data["Order_Date"] = pd.to_datetime(data["Order_Date"])
data["Month"] = data["Order_Date"].dt.month
data["Day"] = data["Order_Date"].dt.day
data["Day_of_Week"] = data["Order_Date"].dt.day_name()

data[["Order_Date", "Month", "Day", "Day_of_Week"]].head()
```

9. **Use apply() to create a useful column**

```python
data["Total_Amount"] = data.apply(
    lambda row: row["Quantity"] * row["Unit_Price"], axis=1
)

data.head()
```

10. **Organize the final DataFrame**

```python
final_data = data[
    [
        "Order_ID",
        "Order_Date",
        "Month",
        "Day",
        "Day_of_Week",
        "Customer_ID",
        "Customer_Name",
        "City",
        "Region",
        "Membership_Type",
        "Product_ID",
        "Product_Name",
        "Category",
        "Brand",
        "Quantity",
        "Unit_Price",
        "Total_Amount",
        "Payment_Method",
        "Order_Status"
    ]
]

final_data.head()
```

11. **Check the processed dataset**

```python
print("Rows:", final_data.shape[0])
print("Columns:", final_data.shape[1])
final_data.info()
```

12. **Export as CSV**

```python
final_data.to_csv("processed_ecommerce_dataset.csv", index=False)
print("Processed dataset saved successfully!")
```

1. **Title**

Day 10 - Flight Operations Data Analysis

This notebook analyzes the Flight Operations dataset using Python and Pandas.
It covers data inspection, selection, filtering, sorting, grouping,
aggregation, transformations, and key observations.

2. **Import Pandas**

```python
import pandas as pd
```

3. **Load Dataset**

```python
df = pd.read_csv("Day10_Flight_Operations_Dataset.csv")
df.head()
```

4. **Understand the Dataset**

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.info()
df.describe()
```

5. **Check Missing Values**

```python
df.isnull().sum()
```

6. **Data Selection**

```python
df[["Flight_ID", "Airline", "Origin", "Destination", "Passengers"]].head()
```

```python
df[["Flight_ID", "Airline", "Average_Ticket_Price", "Flight_Status"]].head(10)
```

7. **Filtering**

```python
df[df["Passengers"] > 150]
```

```python
df[df["Delay_Minutes"] > 30]
```

```python
df[df["Flight_Status"] == "Cancelled"]
```

```python
df[df["Travel_Class"] == "Business"]
```

8. **Sorting**

```python
df.sort_values("Average_Ticket_Price", ascending=False).head(10)
```

```python
df.sort_values("Delay_Minutes", ascending=False).head(10)
```

```python
df.sort_values("Passengers", ascending=False).head(10)
```

9. **Basic Statistics**

```python
print("Total Flights:", df["Flight_ID"].nunique())
print("Total Passengers:", df["Passengers"].sum())
print("Average Passengers:", df["Passengers"].mean())
print("Average Ticket Price:", df["Average_Ticket_Price"].mean())
print("Highest Ticket Price:", df["Average_Ticket_Price"].max())
print("Lowest Ticket Price:", df["Average_Ticket_Price"].min())
print("Average Delay:", df["Delay_Minutes"].mean())
```

10. **Airline-wise Analysis**

```python
airline_analysis = df.groupby("Airline").agg({
    "Flight_ID": "count",
    "Passengers": "sum",
    "Delay_Minutes": "mean",
    "Passenger_Satisfaction": "mean"
})

airline_analysis
```

```python
airline_analysis.sort_values("Passengers", ascending=False)
```

11. **Flight Status Analysis**

```python
df.groupby("Flight_Status").agg({
    "Flight_ID": "count",
    "Delay_Minutes": "mean",
    "Passenger_Satisfaction": "mean"
})
```

12. **Travel Class Analysis**

```python
df.groupby("Travel_Class").agg({
    "Passengers": "sum",
    "Average_Ticket_Price": "mean",
    "Passenger_Satisfaction": "mean"
})
```

13. **Weather Analysis**

```python
df.groupby("Weather").agg({
    "Flight_ID": "count",
    "Delay_Minutes": "mean",
    "Passenger_Satisfaction": "mean"
}).sort_values("Delay_Minutes", ascending=False)
```

14. **Booking Channel Analysis**

```python
df.groupby("Booking_Channel").agg({
    "Flight_ID": "count",
    "Average_Ticket_Price": "mean",
    "Passenger_Satisfaction": "mean"
})
```

15. **Origin-wise Analysis**

```python
df.groupby("Origin")["Passengers"].sum().sort_values(ascending=False)
```

16. **Destination-wise Analysis**

```python
df.groupby("Destination")["Passengers"].sum().sort_values(ascending=False)
```

17. **Useful Transformation**

```python
df["Delay_Category"] = df["Delay_Minutes"].apply(
    lambda x: "No Delay" if x == 0 else "Delayed"
)

df[["Flight_ID", "Delay_Minutes", "Delay_Category"]].head()
```

```python
df["Delay_Category"].value_counts()
```

18. **Top Performing Airlines**

```python
df.groupby("Airline")["Passenger_Satisfaction"].mean().sort_values(ascending=False)
```

19. **High-Value Flights**

```python
df[df["Average_Ticket_Price"] > 10000][
    ["Flight_ID", "Airline", "Travel_Class", "Average_Ticket_Price"]
].sort_values("Average_Ticket_Price", ascending=False)
```

20. **Observations**


* The dataset gives a good overall picture of flight operations, covering airlines, routes, passengers, ticket prices, delays, weather, and passenger satisfaction.

* Most of the flights in the dataset were completed on time, while delayed flights were also quite common. A smaller portion of flights were cancelled.

* Vistara appears frequently in the dataset, making it one of the more represented airlines.

* Economy class dominates the dataset, showing that most of the recorded passengers travelled in this class.

* Weather seems to have some connection with flight delays, with rainy conditions showing noticeably higher average delays.

* Passenger satisfaction varies across airlines, which suggests that factors beyond just flight status may affect the overall passenger experience.

* Online travel portals are used more often than the other booking channels in this dataset.

* The analysis shows that flight delays, passenger volume, travel class, and booking methods can provide useful insights into overall flight operations.
